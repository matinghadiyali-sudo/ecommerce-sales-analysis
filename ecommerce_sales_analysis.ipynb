```python
df.isnull().sum()
```

```python
#converting Order Date into date type object
```

```python
df['Order_Date']=pd.to_datetime(df['Order_Date'])
df['Order_Date'].dtype
```

```python
#adding months,ays,and year to find trends over them
```

```python
df['Day_Name']=df['Order_Date'].dt.day_name()
df['Month']=df['Order_Date'].dt.month
df['Year']=df['Order_Date'].dt.year
```

```python
#Finding the outlier in the Discount column for the determining the profit and sales over that   particular discount 
```

```python
Q1=df['Discount'].quantile(0.25)
Q3=df['Discount'].quantile(0.75)
IQR=Q3-Q1
def Func(x):
    if(x >= IQR*1.5 + Q3 or x <= Q1-IQR*1.5 ):
        return "Outlier"
    else:
        return "Normal"

df["Outliers"]=df["Discount"].apply(Func)
print(df['Outliers'])
```

```python
#Finding the outlier in Profit
```

```python
Q1=df['Profit'].quantile(0.25)
Q3=df['Profit'].quantile(0.75)
IQR=Q3-Q1
def Func(x):
    if(x >= IQR*1.5 + Q3 or x <=  Q1-IQR*1.5 ):
        return "Outlier"
    else:
        return "Normal"

df["Outliers_Profit"]=df["Profit"].apply(Func)
print(df['Outliers_Profit'])
```

```python
total_sales=sum(df['Sales'])
print(total_sales)
total_profit=sum(df['Profit'])
print(total_profit)
total_order=df['Order_ID'].count()
print(total_order)
```
